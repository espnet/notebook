## One voice

[`kan-bayashi/ljspeech_vits`](https://huggingface.co/espnet/kan-bayashi_ljspeech_vits)
is VITS trained on LJSpeech: one reader, English, 22.05 kHz. `from_pretrained`
downloads it once and caches it.


```python
from espnet2.bin.tts_inference import Text2Speech
from IPython.display import Audio, display

tts = Text2Speech.from_pretrained("espnet/kan-bayashi_ljspeech_vits")

wav = tts("The sale of the hotels is part of a long term strategy.")["wav"]
display(Audio(wav.view(-1).cpu().numpy(), rate=tts.fs))
```

Change the sentence and run the cell again — the model is already loaded, so
it answers in about a second.


## A hundred and twenty-eight voices

[`kan-bayashi/vctk_multi_spk_vits`](https://huggingface.co/espnet/kan-bayashi_vctk_multi_spk_vits)
is the same architecture trained on VCTK, and takes a speaker id. The model
says how many it has, so nothing here has to be told.


```python
import numpy as np

multi = Text2Speech.from_pretrained("espnet/kan-bayashi_vctk_multi_spk_vits")
print(f"{multi.tts.spks} speakers")

for speaker in (0, 7, 42):
    wav = multi("Every voice here is a different person.",
                sids=np.array([speaker]))["wav"]
    print(f"speaker {speaker}")
    display(Audio(wav.view(-1).cpu().numpy(), rate=multi.fs))
```
